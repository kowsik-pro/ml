# Post Lab: Exploratory Data Analysis on Titanic Dataset

## 1. Import Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df=sns.load_dataset("titanic")
df.head()
```

## 2. Dataset Information

```python
print(df.shape)
print(df.columns)
df.info()
```

## 3. Statistical Summary

```python
df.describe().transpose()
```

## 4. Missing Values

```python
df.isnull().sum()
```

## 5. Missing Value Heatmap

```python
plt.figure(figsize=(8,5));sns.heatmap(df.isnull(),yticklabels=False,cbar=False,cmap='viridis');plt.title(f'Missing Values(24EU01110)');plt.show()
```

## 6. Survival Count

```python
sns.countplot(x='survived',data=df)
plt.title(f'Survival Count(24EU01110)')
plt.show()
```

## 7. Gender vs Survival

```python
sns.countplot(x='sex',hue='survived',data=df)
plt.title(f'Gender vs Survival(24EU01110)')
plt.show()
```

## 8. Pclass vs Survival

```python
sns.countplot(x='pclass',hue='survived',data=df)
plt.title(f'Pclass vs Survival(24EU01110)')
plt.show()
```

## 9. Age Distribution

```python
sns.histplot(df['age'].dropna(),bins=30,kde=True)
plt.title(f'Age Distribution(24EU01110)')
plt.show()
```

## 10. Fare Distribution

```python
sns.histplot(df['fare'],bins=30,kde=True)
plt.title(f'Fare Distribution(24EU01110)')
plt.show()
```

## 11. Fare Boxplot

```python
sns.boxplot(x=df['fare'])
plt.title(f'Fare Boxplot(24EU01110)')
plt.show()
```

## 12. Correlation Heatmap

```python
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap='coolwarm')
plt.title(f'Correlation Heatmap(24EU01110)')
plt.show()
```

## 13. Observations

```python
print("1. Female passengers had higher survival.\n2. First class had better survival.\n3. Age contains missing values.\n4. Fare has outliers.")
```
